2. WEFAT

```python
import numpy as np
from scipy.spatial.distance import cosine
# Define words to test
targets = ['cina', 'tionghoa', 'pribumi', 'asli']
positive = ['sukses', 'pintar', 'pekerja keras', 'kaya', 'modern']
negative = ['asing', 'serakah', 'monopoli', 'pendatang', 'tidak loyal']
# This will likely show which religion is more associated with extremism in news

# Calculate bias
for target in targets:
    target_vec = model.get_word_vector(target)
    
    pos_sim = np.mean([1 - cosine(target_vec, model.get_word_vector(w)) for w in positive])
    neg_sim = np.mean([1 - cosine(target_vec, model.get_word_vector(w)) for w in negative])
    bias = pos_sim - neg_sim
    
    print(f"{target}: bias score = {bias:.4f} (positive={pos_sim:.4f}, negative={neg_sim:.4f})")
```
